```python
import json
from kafka import KafkaConsumer

KAFKA_BROKER = "192.168.137.119"  # Mengarah ke IP PC Master
TOPIC_NAME = "wikimedia-recentchange"

consumer = KafkaConsumer(
    TOPIC_NAME,
    bootstrap_servers=[KAFKA_BROKER],
    group_id="uji-decoupling-group",
    auto_offset_reset="earliest",
    enable_auto_commit=True,
    auto_commit_interval_ms=1000,
    value_deserializer=lambda m: json.loads(m.decode("utf-8"))
)

kunci_terbaca = set()
total_diterima = 0
jumlah_duplikat = 0

print("Consumer di Worker aktif... Tekan Ctrl+C untuk simulasi mati sementara.")
try:
    for msg in consumer:
        event = msg.value
        unique_id = f"{event.get('wiki')}_{event.get('id')}"
        total_diterima += 1

        if unique_id in kunci_terbaca:
            jumlah_duplikat += 1
        else:
            kunci_terbaca.add(unique_id)

        if total_diterima % 200 == 0:
            print(f"Partisi {msg.partition} | Offset {msg.offset} | Total dibaca: {total_diterima} | Duplikat sesi ini: {jumlah_duplikat}")
except KeyboardInterrupt:
    print(f"\nConsumer Worker dihentikan. Total dibaca: {total_diterima}, Duplikat: {jumlah_duplikat}")
finally:
    consumer.close()
```
